# Gradient Descent Optimization

**Objective:** Use gradient descent to find the minimum of the function \(f(x)=(x+3)^2\), and visualize the optimization steps.

## 1. Import libraries

```python
import numpy as np
import matplotlib.pyplot as plt
```

## 2. Define the function and its derivative

```python
def function(x):
    return (x + 3) ** 2

def derivative(x):
    return 2 * (x + 3)
```

## 3. Apply gradient descent

```python
x = 2
learning_rate = 0.1
iterations = 50

x_values = []
y_values = []

for i in range(iterations):
    x_values.append(x)
    y_values.append(function(x))
    gradient = derivative(x)
    x = x - learning_rate * gradient

print("Local minimum x:", round(x, 6))
print("Minimum value y:", round(function(x), 6))
```

## 4. Plot the function and optimization steps

```python
x_range = np.linspace(-5, 4, 100)
y_range = function(x_range)

plt.plot(x_range, y_range, label="y = (x + 3)^2")
plt.scatter(x_values, y_values, color="red", label="Gradient Descent Steps")
plt.scatter(x, function(x), color="green", label="Minimum")
plt.xlabel("x")
plt.ylabel("y")
plt.title("Gradient Descent")
plt.legend()
plt.grid(True)
plt.show()
```

## Conclusion

Gradient descent repeatedly updates \(x\) in the direction opposite to the derivative. For this function, the minimum is at \(x=-3\), where \(f(x)=0\). With a learning rate of 0.1 and 50 iterations, the algorithm approaches this minimum.
